```python
import sys
sys.path.append('../src')

from preprocessing import preprocess
from train_test_split import get_train_test_split
from train_model import scale_features, train_baseline_model
from evaluate import evaluate_model

df = preprocess('../data/raw/telco_churn.csv')
X_train, X_test, y_train, y_test = get_train_test_split(df)
X_train_scaled, X_test_scaled, scaler = scale_features(X_train, X_test)

model = train_baseline_model(X_train_scaled, y_train)
evaluate_model(model, X_test_scaled, y_test)
```

```python
from train_model import train_random_forest
from evaluate import evaluate_model

rf_model = train_random_forest(X_train, y_train)  # unscaled — trees don't need scaling
evaluate_model(rf_model, X_test, y_test)
```

```python
from train_model import save_model

save_model(rf_model, '../models/churn_rf_model.pkl')
```

```python
from train_model import load_model

loaded_model = load_model('../models/churn_rf_model.pkl')

# Sanity check: does the loaded model give the same predictions as the original?
import numpy as np
original_preds = rf_model.predict(X_test)
loaded_preds = loaded_model.predict(X_test)

print(np.array_equal(original_preds, loaded_preds))
```
